## 4. Preprocesamiento de datos

Carga los CSVs, elimina la fila 'Total', convierte columnas de porcentaje y llena valores faltantes.

````python
BUQUES_CSV = f"{BASE_DIR}/Trafico_Buques_Segmento_Mercado_Canal_Panama.csv"
PAISES_CSV = f"{BASE_DIR}/Principales_Paises_Flujo_Carga_Canal_Panama.csv"

# Cargar y preprocesar datos de buques
buques = pd.read_csv(BUQUES_CSV)
buques = buques[buques["Segmento de Mercado"] != "Total"].copy()
buques["% Diferencia Tránsitos"] = (
    buques["% Diferencia Tránsitos"]
    .str.replace("%", "", regex=False)
    .str.strip()
    .astype(float)
)
buques["Toneladas Largas de Carga 2024"] = buques["Toneladas Largas de Carga 2024"].fillna(0)
buques["Toneladas Largas de Carga 2025"] = buques["Toneladas Largas de Carga 2025"].fillna(0)
buques["Cambio"] = buques["Tránsitos 2025"] - buques["Tránsitos 2024"]

# Cargar datos de países
paises = pd.read_csv(PAISES_CSV)
paises["Costa a Costa"] = paises["Costa a Costa"].fillna(0)

print("Datos de buques:")
display(buques)
print("\nDatos de países:")
display(paises)
````

## 5. KPIs principales

Calcula las métricas clave del análisis.

````python
# Entrenar modelo de predicción
X = buques[["Tránsitos 2024"]].values
y = buques["Tránsitos 2025"].values
model = LinearRegression()
model.fit(X, y)

# Predicción 2026
X_2026 = buques[["Tránsitos 2025"]].values
preds_2026 = model.predict(X_2026)

# Métricas principales
total_2024 = int(buques["Tránsitos 2024"].sum())
total_2025 = int(buques["Tránsitos 2025"].sum())
variacion = (total_2025 - total_2024) / total_2024 * 100

crecimiento = buques.loc[buques["Cambio"].idxmax()]
caida = buques.loc[buques["Cambio"].idxmin()]
pais_top = paises.loc[paises["Total"].idxmax()]
pred_total_2026 = int(preds_2026.sum())

print(f"Tránsitos 2024: {total_2024:,}")
print(f"Tránsitos 2025: {total_2025:,}")
print(f"Variación: {variacion:.2f}%")
print(f"Mayor crecimiento: {crecimiento['Segmento de Mercado']} (+{int(crecimiento['Cambio'])})")
print(f"Mayor caída: {caida['Segmento de Mercado']} ({int(caida['Cambio'])})")
print(f"Top país: {pais_top['País']} ({pais_top['Total']/1e6:.1f}M ton)")
print(f"Predicción 2026: {pred_total_2026:,} tránsitos")
````

## 6. Gráfico de tránsitos por segmento

Compara los tránsitos de 2024 vs 2025 para cada segmento de mercado.

````python
fig = go.Figure()
x = range(len(buques))
fig.add_trace(go.Bar(x=list(x), y=buques["Tránsitos 2024"], name="2024", marker_color="steelblue"))
fig.add_trace(go.Bar(x=list(x), y=buques["Tránsitos 2025"], name="2025", marker_color="seagreen"))
fig.update_layout(
    title="Tránsitos 2024 vs 2025 por Segmento de Mercado",
    xaxis_tickangle=45,
    xaxis_ticktext=buques["Segmento de Mercado"].tolist(),
    yaxis_title="Número de tránsitos",
    barmode="group",
    height=500,
)
fig.show()
````

## 7. Gráfico de variación porcentual

Muestra el porcentaje de cambio en tránsitos por segmento (verde = crecimiento, rojo = caída).

````python
colors = ["green" if c > 0 else "red" for c in buques["Cambio"]]
fig2 = go.Figure()
fig2.add_trace(go.Bar(
    x=buques["Segmento de Mercado"],
    y=buques["% Diferencia Tránsitos"],
    marker_color=colors,
))
fig2.update_layout(
    title="% Diferencia Tránsitos por Segmento",
    xaxis_tickangle=45,
    yaxis_title="Variación %",
    height=500,
)
fig2.show()
````

## 8. Mapa de calor por país

Visualiza el volumen de carga total por país usando un mapa coroplético.

````python
COUNTRY_ISO3 = {
    "Estados Unidos": "USA",
    "China": "CHN",
    "Japón": "JPN",
    "Chile": "CHL",
    "Corea del Sur": "KOR",
    "Perú": "PER",
    "México": "MEX",
    "Colombia": "COL",
    "Ecuador": "ECU",
    "Canadá": "CAN",
    "Panamá": "PAN",
    "Brasil": "BRA",
    "España": "ESP",
    "Guatemala": "GTM",
    "Holanda (Países Bajos)": "NLD",
}

paises_map = paises.copy()
paises_map["ISO3"] = paises_map["País"].map(COUNTRY_ISO3)
paises_map = paises_map.dropna(subset=["ISO3"])

fig3 = go.Figure(go.Choropleth(
    locations=paises_map["ISO3"],
    z=paises_map["Total"],
    text=paises_map["País"],
    colorscale="Blues",
    colorbar_title="Toneladas",
))
fig3.update_layout(title="Volumen de Carga por País", height=500, margin=dict(l=0, r=0, t=40, b=0))
fig3.show()
````

## 9. Top países

Ranking de los 10 países con mayor volumen de carga.

````python
top10 = paises.nlargest(10, "Total")
fig4 = go.Figure(go.Bar(
    y=top10["País"][::-1],
    x=top10["Total"][::-1] / 1e6,
    orientation="h",
    marker_color="steelblue",
))
fig4.update_layout(title="Top 10 Países por Volumen de Carga", xaxis_title="Millones de toneladas", height=500)
fig4.show()
````

## 10. Predicción 2026

Utiliza regresión lineal para predecir los tránsitos de 2026 basándose en la relación 2024-2025.

````python
buques_pred = buques.copy()
buques_pred["Predicción 2026"] = preds_2026.round(0).astype(int)

fig5 = go.Figure()
x = range(len(buques_pred))
fig5.add_trace(go.Bar(x=list(x), y=buques_pred["Tránsitos 2024"], name="2024", marker_color="steelblue"))
fig5.add_trace(go.Bar(x=list(x), y=buques_pred["Tránsitos 2025"], name="2025", marker_color="seagreen"))
fig5.add_trace(go.Bar(x=list(x), y=buques_pred["Predicción 2026"], name="2026 (predicción)", marker_color="orange"))
fig5.update_layout(
    title="Predicción 2026 por Segmento de Mercado",
    xaxis_tickangle=45,
    xaxis_ticktext=buques_pred["Segmento de Mercado"].tolist(),
    yaxis_title="Número de tránsitos",
    barmode="group",
    height=500,
)
fig5.show()

print(f"\nPredicción total 2026: {pred_total_2026:,} tránsitos")
print("Nota: Esta predicción es una estimación lineal simple y no constituye una garantía.")
````

## 11. Resumen Ejecutivo con IA

Genera un resumen ejecutivo utilizando IA con tres niveles de prioridad:

1. **OpenRouter** (requiere API key de OpenRouter)
2. **Ollama local** (requiere Ollama corriendo en localhost:11434 con un modelo cargado, ej: `ollama run llama3.2`)
3. **Resumen automático** (sin LLM, generación basada en reglas)

### Configuración

Para usar **Ollama local**:

### Ejecutar la celda

La función `get_llm_summary()` detecta automáticamente qué servicio está disponible.

````python
# =============================================================================
# RESUMEN EJECUTIVO CON IA
# =============================================================================
# Orden de prioridad:
# 1. OpenRouter (requiere API key)
# 2. Ollama local (requiere Ollama corriendo en localhost:11434)
# 3. Resumen automático sin LLM

def is_ollama_available():
    try:
        requests.get("http://localhost:11434/api/tags", timeout=2)
        return True
    except:
        return False

def get_llm_summary(kpis_text, api_key=None, ollama_model="llama3.2"):
    # 1. Intentar OpenRouter
    if api_key:
        try:
            from openai import OpenAI
            client = OpenAI(api_key=api_key, base_url="https://openrouter.ai/api/v1")
            response = client.chat.completions.create(
                model="nvidia/nemotron-3-ultra-550b-a55b:free",
                messages=[{"role": "user", "content": f"Resume en español en máximo 120 palabras. Incluye KPIs y advertencia del modelo: {kpis_text}"}],
                max_tokens=250,
            )
            return response.choices[0].message.content, "OpenRouter"
        except Exception as e:
            print(f"OpenRouter no disponible: {e}")
    
    # 2. Intentar Ollama
    if is_ollama_available():
        try:
            resp = requests.post(
                "http://localhost:11434/api/generate",
                json={"model": ollama_model, "prompt": f"Resume en español en máximo 120 palabras: {kpis_text}", "stream": False},
                timeout=30
            )
            if resp.status_code == 200:
                return resp.json()["response"], "Ollama"
        except Exception as e:
            print(f"Ollama no disponible: {e}")
    
    # 3. Fallback automático
    return auto_summary, "Automático"


auto_summary = (
    f"En 2025 el Canal de Panamá registró {total_2025:,} tránsitos, "
    f"un {variacion:.2f}% más que los {total_2024:,} de 2024. "
    f"El segmento de {crecimiento['Segmento de Mercado']} fue el de mayor crecimiento absoluto "
    f"({int(crecimiento['Cambio'])} barcos más). "
    f"Por volumen de carga, {pais_top['País']} lidera el ranking. "
    f"El modelo lineal predice {pred_total_2026:,} tránsitos para 2026. "
    f"Esta predicción es una estimación simplificada y no constituye una garantía."
)

kpis_text = (
    f"Tránsitos 2024: {total_2024:,} | "
    f"Tránsitos 2025: {total_2025:,} | "
    f"Variación: {variacion:.2f}% | "
    f"Mayor crecimiento: {crecimiento['Segmento de Mercado']} (+{int(crecimiento['Cambio'])}) | "
    f"Mayor caída: {caida['Segmento de Mercado']} ({int(caida['Cambio'])}) | "
    f"Top país: {pais_top['País']} ({pais_top['Total']/1e6:.1f}M ton) | "
    f"Predicción 2026: {pred_total_2026:,} tránsitos"
)

# Configura tu API key aquí (o déjala como None para usar Ollama o el resumen automático)
api_key = None  # Cambia a tu key de OpenRouter si tienes una
ollama_model = "llama3.2"

summary, source = get_llm_summary(kpis_text, api_key=api_key, ollama_model=ollama_model)
print(f"Resumen generado con: {source}")
print(f"\n{summary}")
````
